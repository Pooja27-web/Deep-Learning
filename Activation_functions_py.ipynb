```python
import numpy as np
import matplotlib.pyplot as plt

# Sigmoid: squashes input to a range between 0 and 1
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# ReLU: outputs the input directly if positive, otherwise outputs 0
def relu(x):
    return np.maximum(0, x)

# Tanh: squashes input to a range between -1 and 1
def tanh(x):
    return np.tanh(x)
```

```python
inputs = np.array([-5, -2, -1, 0, 1, 2, 5], dtype=float)

sigmoid_outputs = sigmoid(inputs)
relu_outputs = relu(inputs)
tanh_outputs = tanh(inputs)

print(f"{'Input':>8} | {'Sigmoid':>10} | {'ReLU':>8} | {'Tanh':>10}")
print("-" * 45)
for x, s, r, t in zip(inputs, sigmoid_outputs, relu_outputs, tanh_outputs):
    print(f"{x:8.1f} | {s:10.4f} | {r:8.1f} | {t:10.4f}")
```

```python
x_range = np.linspace(-10, 10, 400)

plt.figure(figsize=(8, 5))
plt.plot(x_range, sigmoid(x_range), label="Sigmoid", color="#1f77b4", linewidth=2)
plt.plot(x_range, relu(x_range), label="ReLU", color="#2ca02c", linewidth=2)
plt.plot(x_range, tanh(x_range), label="Tanh", color="#d62728", linewidth=2)
plt.axhline(0, color='gray', linewidth=0.6)
plt.axvline(0, color='gray', linewidth=0.6)
plt.title("Comparison of Activation Functions")
plt.xlabel("Input (x)")
plt.ylabel("Output f(x)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("activation_functions_plot.png", dpi=200)
```
